```python
import numpy as np
import plotly.graph_objects as go
from scipy.stats import multivariate_normal

# Mean and covariance matrix
mean = [0, 0]
cov = [[1, 0], [0, 1]]

# Create a meshgrid
x, y = np.meshgrid(np.linspace(-5, 5, 1000), np.linspace(-5, 5, 1000))

# Create a bivariate normal distribution object
rv = multivariate_normal(mean, cov)

def f(x, y):
    return (x**2 - y**2) / (x**2 + y**2)

# Evaluate the PDF at the points in the meshgrid
z = f(x, y)

# Create a 3D surface plot using Plotly
fig = go.Figure(data=[go.Surface(z=z, x=x, y=y, colorscale='viridis')])
fig.update_layout(title='Bivariate Normal Distribution',
                  scene=dict(xaxis_title='X',
                             yaxis_title='Y',
                             zaxis_title='Probability Density'))

# Show the interactive plot
fig.show(renderer="browser")
```
