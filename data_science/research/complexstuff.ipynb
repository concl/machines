```python
import numpy as np
import plotly.graph_objects as go
from scipy.special import gamma

# Complex 2D Grid
x = np.linspace(-10, 10, 400)
y = np.linspace(-10, 10, 400)
X, Y = np.meshgrid(x, y)
Z = X + 1j * Y

# Compute Complex Gamma Function & Complex Logarithm
gamma_values = gamma(Z)
log_values = np.log(Z)

def plot_complex_function_3d(X, Y, func_values, title):
    real_part = np.real(func_values)
    imag_part = np.imag(func_values)

    fig = go.Figure(data=[go.Surface(z=real_part, x=X, y=Y, surfacecolor=imag_part, colorscale='viridis', cmin=-5, cmax=5)])
    fig.update_layout(title=title,
                      scene=dict(xaxis_title='Re(z)',
                                 yaxis_title='Im(z)',
                                 zaxis_title='Real Part of Value'),
                      coloraxis_colorbar=dict(title="Imaginary Part"))
    fig.show(renderer="browser")

# Plotting the functions
plot_complex_function_3d(X, Y, gamma_values, "Gamma Function")
plot_complex_function_3d(X, Y, log_values, "Complex Logarithm")
```
