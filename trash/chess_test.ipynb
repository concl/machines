```python
import chess.engine

stockfish_path = "C:/Users/Luke Zeng/Downloads/stockfish-windows-x86-64-avx2/stockfish/stockfish-windows-x86-64-avx2.exe"  # Replace with the path to your Stockfish binary
board = chess.Board()

with chess.engine.SimpleEngine.popen_uci(stockfish_path) as engine:
    while True:
        result = engine.play(board, chess.engine.Limit(time=2.0))  # Play a move with 2 seconds thinking time
        board.push(result.move)
        print(board)
        if board.is_game_over():
            break

        next_move = input("Enter your move: ")
        board.push_san(next_move)
        print(board)


```
