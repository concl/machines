```python
import numpy as np
import matplotlib.pyplot as plt
from plotly import graph_objects as go
from einops import rearrange
```

```python
np.set_printoptions(precision=3)
```

```python

def quadratic_function(x):
    Q = np.array([[3, 1], [1, 2]])
    b = np.array([[1, 1]])
    c = 0.5
    
    return 0.5 * np.matmul(x.swapaxes(-1, -2), np.matmul(Q, x)) + np.matmul(b, x) + c

num_x_points = 16
num_y_points = 16

x, y = np.meshgrid(np.linspace(-5, 5, num_x_points), np.linspace(-5, 5, num_y_points), indexing='xy')
x, y = rearrange(x, 'h w -> h w 1'), rearrange(y, 'h w -> h w 1')
coordinates = np.concatenate([x, y], axis=-1)
coordinates = rearrange(coordinates, 'h w c -> h w c 1')
z = quadratic_function(coordinates)
z = z.reshape(num_x_points, num_y_points)

```

## Matplotlib 3D plot

```python
%matplotlib widget
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(222, projection='3d')
ax.plot_surface(x.squeeze(), y.squeeze(), z, cmap='viridis')
ax.set_xlabel('X axis')
ax.set_ylabel('Y axis')
ax.set_zlabel('Z axis')
ax.set_title('3D Plot of Quadratic Function')
plt.show()


```

## plotly 3D plot

```python

sample_data = np.random.rand(100, 3)

fig = go.Figure(data=[go.Surface(z=z, x=x.squeeze(), y=y.squeeze(), colorscale='Viridis')])
fig.add_trace(go.Scatter3d(x=sample_data[:, 0], y=sample_data[:, 1], z=sample_data[:, 2],
                           mode='markers', marker=dict(size=5, color='red')))
fig.update_layout(title='3D Plot of Quadratic Function', autosize=True,
                  scene=dict(xaxis_title='X axis', yaxis_title='Y axis', zaxis_title='Z axis'))
```
