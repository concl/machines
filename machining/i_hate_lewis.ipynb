```python
from typing import Union

x = np.linspace(0, 99, 100)

def integral(f, a, b, n):

    # using simpson's rule
    h = (b - a) / n
    s = f(a) + f(b)

    for i in range(1, n, 2):
        s += 4 * f(a + i * h)
    for i in range(2, n-1, 2):
        s += 2 * f(a + i * h)
    
    return s * h / 3


def f(x):
    return sin(x)

print(integral(f, 0, 10, 1000))



```
